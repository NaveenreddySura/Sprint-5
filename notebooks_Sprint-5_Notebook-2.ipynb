# Data Type Handling

## Identifying Data Types
Definition
Identifying data types means determining what kind of information each column contains.
Data may represent numbers, categories, text, Boolean values, or dates.
Understanding data types helps select appropriate preprocessing, analysis, and machine learning techniques.

#### Business example
An e-commerce company stores:
- Customer Name → String
- Product Price → Float
- Quantity → Integer
- Order Date → DateTime
- Payment Completed → Boolean
#### AI/ML use case
Before building a customer prediction model, a data scientist identifies which columns are numerical, categorical, and date-based.
This helps determine which columns require scaling, encoding, or conversion.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 30, 35],
    "Salary": [30000.5, 40000.0, 50000.5],
    "Purchased": [True, False, True]
})

print(df.dtypes)

print("\nData Types:")
print(df.info())
```

#### Explanation
- dtypes displays the data type of every column.
- info() displays column names, data types, and non-null counts.
- Age is stored as an integer.
- Salary is stored as a floating-point number.
- Purchased is stored as Boolean.
- Name is stored as object/string-like data.

## Numerical Data
Definition
Numerical data represents information using numbers that can be used for mathematical calculations.
It includes integers and floating-point numbers.
Numerical data can be further classified into discrete and continuous data.
#### Real-world example
- Number of students → 50
- Height → 175.5 cm
- Weight → 60.5 kg
- Age → 25 years
#### Business example
A company analyzes:
- Product price
- Revenue
- Sales quantity
- Employee salary
- Customer spending
These values are numerical and can be used for calculations.
#### AI/ML use case
A house price prediction model uses house area, number of bedrooms, and house age as numerical features.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [25, 30, 35],
    "Salary": [30000.5, 40000.5, 50000.5]
})

print(df["Age"].mean())
print(df["Salary"].mean())

print("\nData Types:")
print(df.dtypes)
```

#### Explanation
- mean() calculates the average.
- Age is an integer column.
- Salary is a float column.
- Numerical columns support mathematical operations.

## Categorical Data
Definition
Categorical data represents information divided into groups, labels, or categories.
It usually describes qualities or classifications rather than measurable quantities.
Categorical data can be nominal or ordinal.
#### Real-world example
- Gender → Male, Female
- City → Hyderabad, Bengaluru
- Blood Group → A, B, AB, O
- Education Level → School, Degree, Masters
#### Business example
A company categorizes customers according to:
- Department
- Product category
- Payment method
- Customer segment
- Order status
#### AI/ML use case
A machine learning model predicts customer purchases using categorical features such as gender, city, and product category.

These categories may need to be encoded into numerical representations.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "IT"],
    "Payment": ["UPI", "Card", "Cash", "UPI"]
})

print(df["Department"].value_counts())

print("\nUnique Departments:")
print(df["Department"].unique())

print("\nData Types:")
print(df.dtypes)
```

#### Explanation
- value_counts() counts the frequency of each category.
- unique() identifies distinct categories.
- Department and Payment are categorical columns.
- Pandas generally stores these columns as object unless explicitly converted.

## Boolean Data
Definition
Boolean data represents logical values with only two possible states: True or False.
It is commonly used to represent conditions, flags, and yes/no information.
Boolean values are useful for filtering data and checking conditions.
#### Real-world example
- Is the student present? → True
- Is the exam passed? → False
- Is the payment completed? → True
#### Business example
An online shopping system stores:
- Is the order delivered?
- Is the payment successful?
- Is the customer active?
- Is the product available?
These fields can be represented using Boolean values.
#### AI/ML use case
A customer churn dataset may contain a Boolean column indicating whether a customer has subscribed to a premium service.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["Ravi", "Priya", "Arun"],
    "Active": [True, False, True]
})

print(df)

print("\nActive Customers:")
print(df[df["Active"] == True])
```

#### Explanation
- Active contains Boolean values.
- True represents active customers.
- False represents inactive customers.
- Boolean conditions are used to filter records.

## Date/Time Data
Definition
Date/time data represents dates, times, or timestamps.
It is useful for understanding when events happened and calculating time differences.
Pandas uses the datetime64 data type for many date/time operations.
#### Real-world example
- Date of Birth → 2000-05-15
- Joining Date → 2023-07-10
- Appointment Time → 2026-11-05 10:30:00
#### Business example
An e-commerce company analyzes:
- Order date
- Delivery date
- Payment date
- Customer registration date
This helps calculate delivery duration and monthly sales.
#### AI/ML use case
A sales forecasting model uses transaction dates to identify monthly, weekly, and seasonal purchasing patterns.

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Date": [
        "2026-01-10",
        "2026-02-15",
        "2026-03-20"
    ]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print(df)

print("\nData Type:")
print(df.dtypes)

print("\nMonth:")
print(df["Order_Date"].dt.month)

print("\nYear:")
print(df["Order_Date"].dt.year)
```

#### Explanation
- pd.to_datetime() converts strings into date/time values.
- .dt.month extracts the month.
- .dt.year extracts the year.
- Date columns allow time-based analysis.

## String Data
Definition
String data represents textual information using characters, words, numbers, or symbols.
It is commonly used to store names, addresses, descriptions, IDs, and other text-based information.
String data can be processed using text operations such as strip(), lower(), upper(), and replace().
#### Real-world example
- Name → "Ravi"
- City → "Kadapa"
- Mobile Number → "9876543210"
- Email → "ravi@gmail.com"
#### Business example
A company stores customer names, email addresses, product descriptions, and shipping addresses as strings.
#### AI/ML use case
A sentiment analysis model processes customer reviews stored as strings to identify positive, negative, or neutral opinions.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": [" Ravi ", "PRIYA", " arun "],
    "City": ["Kadapa", "Hyderabad", "Bengaluru"]
})

df["Customer"] = df["Customer"].str.strip().str.title()

print(df)
```

#### Explanation
- str.strip() removes unnecessary spaces.
- str.title() converts names into title case.
- String operations help standardize textual information.
- This improves consistency in customer records.

## Converting Data Types
Definition
Data type conversion means changing a column from one data type to another.
It is performed when data is stored in an incorrect format or when a particular operation requires a different data type.
Proper conversion allows calculations, comparisons, date operations, and machine learning preprocessing.
#### Real-world example
A student marks column is stored as "85" instead of 85.

Converting it into an integer allows mathematical calculations.
#### Business example
A sales dataset contains product prices stored as text.
The company converts prices into numerical values to calculate revenue and profit.
#### AI/ML use case
A machine learning model cannot directly perform numerical calculations on values stored as ordinary text.
Data type conversion prepares the features for numerical processing.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["25", "30", "35"],
    "Salary": ["30000", "40000", "50000"]
})

print("Before Conversion:")
print(df.dtypes)

df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].astype(float)

print("\nAfter Conversion:")
print(df.dtypes)
```

#### Explanation
- Initially, Age and Salary are stored as strings.
- astype(int) converts Age into integer values.
- astype(float) converts Salary into floating-point values.
- Numerical operations become possible after conversion.

## astype()
Definition
astype() is a Pandas method used to convert a Series or DataFrame column into a specified data type.
It supports conversions such as integer, float, string, Boolean, and categorical data types.
It is useful when values are already valid for the target data type.
#### Real-world example
A student marks column contains:
"80", "90", "75"
Using astype(int) converts these strings into integers.
#### Business example
A company converts product quantities from string format into integers before calculating total inventory.
#### AI/ML use case
A dataset contains customer age values stored as strings.
Converting Age into numerical format allows the model to process age as a numerical feature.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["25", "30", "35"],
    "Salary": ["30000", "40000", "50000"]
})

df["Age"] = df["Age"].astype(int)

df["Salary"] = df["Salary"].astype(float)

df["Salary"] = df["Salary"].astype(int)

print(df)
print(df.dtypes)
```

#### Explanation
- Age is converted into integer.
- Salary is first converted into float.
- Salary is then converted into integer.
- astype(int) removes the decimal portion when converting floats; it does not round the value.

## to_numeric()
Definition
pd.to_numeric() converts values into numerical data types.
It is especially useful when numerical values are stored as strings or when a column contains invalid numerical values.
It supports error handling using the errors parameter.
#### Real-world example
A marks sheet contains:
"85", "90", "Absent", "75"
The text "Absent" cannot be converted directly into a number.
#### Business example
A company receives sales data from Excel where prices contain text values such as "Unknown" or "N/A".
The company converts valid prices into numerical values and handles invalid entries.
#### AI/ML use case
A customer dataset contains income values stored as strings.
Converting income into numeric format allows the model to use income as a numerical feature.

```python
import pandas as pd

df = pd.DataFrame({
    "Marks": ["85", "90", "Absent", "75", "Invalid"]
})

df["Marks"] = pd.to_numeric(
    df["Marks"],
    errors="coerce"
)

print(df)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

#### Explanation
- pd.to_numeric() converts valid values into numbers.
- "85", "90", and "75" become numeric values.
- "Absent" and "Invalid" become NaN.
- errors="coerce" converts invalid values into missing values instead of raising an error.

## to_datetime()
Definition
pd.to_datetime() converts strings or other supported values into date/time format.
It enables date-based calculations, filtering, sorting, and extracting time-related information.
It is useful when dates are stored as strings in CSV, Excel, or database exports.
#### Real-world example
A school stores student joining dates as:
"10-01-2023"
Converting this into datetime allows calculating how long the student has been enrolled.
#### Business example
A company converts order dates into datetime format to analyze monthly revenue and delivery times.
#### AI/ML use case
A forecasting model uses date/time features to identify seasonal patterns in customer purchases.

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Date": [
        "2026-01-10",
        "2026-02-15",
        "2026-03-20"
    ]
})

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"]
)

print(df)

print("\nData Type:")
print(df.dtypes)

print("\nMonth:")
print(df["Order_Date"].dt.month)
```

#### Explanation
- String dates are converted into datetime.
- The column becomes datetime64[ns].
- .dt.month extracts the month.
- Date/time conversion makes time-based calculations possible.

```python
df = pd.DataFrame({
    "Date": ["15-01-2026", "20-02-2026"]
})

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y"
)

print(df)
```

#### Explanation
- %d represents the day.
- %m represents the month.
- %Y represents the four-digit year.
- Specifying the format helps interpret dates correctly.

## Handling Invalid Conversions
Definition
Invalid conversions occur when values cannot be converted into the requested data type.
They may contain unexpected text, incorrect formats, symbols, or unsupported values.
Invalid conversions must be identified and handled appropriately before analysis or model training.
#### Real-world example
A marks column contains:
- "85"
- "90"
- "Absent"
- "Invalid"
Only the first two values are directly convertible into numbers.
#### Business example
A company receives customer income values:
- "50000"
- "60000"
- "Unknown"
- "N/A"
The company must identify invalid values before calculating average income.
#### AI/ML use case
A machine learning model requires numerical income values.
Invalid income entries must be converted into missing values and handled through appropriate processing.

```python
import pandas as pd

df = pd.DataFrame({
    "Income": ["30000", "50000", "Unknown", "N/A", "70000"]
})

df["Income"] = pd.to_numeric(
    df["Income"],
    errors="coerce"
)

print("Converted Data:")
print(df)

print("\nInvalid Values:")
print(df[df["Income"].isnull()])

# Fill missing values
df["Income"] = df["Income"].fillna(
    df["Income"].median()
)

print("\nFinal Data:")
print(df)
```

#### Explanation
- to_numeric() converts valid income values.
- errors="coerce" converts invalid values into NaN.
- isnull() identifies invalid converted values.
- median() calculates the median of valid income values.
- fillna() replaces missing values with the median.

#### Detecting Incorrect Data Types
Definition
Detecting incorrect data types means identifying columns whose stored data types do not match their actual meaning.
For example, a salary column may be stored as text instead of numerical data.
Identifying these issues helps prevent calculation errors and preprocessing problems.
#### Real-world example
A person's age is stored as "25" instead of 25.
Although both represent the same value, the string cannot be directly used in many numerical operations.
#### Business example
A sales dataset contains:
Product	Price
Laptop	"50000"
Mobile	"20000"
Tablet	"15000"

Price is stored as text instead of numerical data.
#### AI/ML use case
A machine learning model receives customer age as a string.

Before training, the data scientist converts Age into a numerical feature.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["25", "30", "35"],
    "Salary": ["30000", "40000", "50000"],
    "City": ["Kadapa", "Hyderabad", "Bengaluru"]
})

print("Current Data Types:")
print(df.dtypes)

# Convert incorrect types
df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])

print("\nCorrected Data Types:")
print(df.dtypes)
```

#### Explanation
- Initially, Age and Salary are stored as strings.
- to_numeric() converts them into numerical columns.
- City remains a string column because it represents categories.
- Correct data types support appropriate analysis.
