# Data Transformation

## Why Transform Data?
Definition
Data transformation is the process of changing the mathematical representation of data without necessarily changing its original meaning. It helps make data easier to analyze and can improve the performance of certain machine learning algorithms.
Transformations are especially useful when numerical data contains skewness, extreme values, or distributions that are difficult for statistical models to handle.
#### Real-World Example
Suppose a company's employee salaries are:
[20000, 25000, 30000, 35000, 200000]
The salary of ₹2,00,000 is much higher than the other salaries. This creates a right-skewed distribution.
A transformation can reduce the influence of this extreme value and make the distribution more balanced.
#### Business Example
An e-commerce company analyzes customer spending.
Most customers spend ₹500–₹2,000, but a few customers spend ₹50,000 or more.
Transforming customer spending can help the company:
- Analyze customer behavior more effectively.
- Reduce extreme differences in numerical values.
- Improve certain statistical models.
- Identify patterns in customer purchasing behavior.
#### AI/ML Use Case
Data transformation can be useful for:
- Linear Regression
- Logistic Regression
- Statistical modeling
- Customer segmentation
- Income prediction
- Sales forecasting

## Understanding Skewness
Definition
Skewness measures how asymmetric a numerical distribution is.
A perfectly symmetrical distribution has approximately equal shapes on both sides of its center. A skewed distribution has a longer tail on one side.

### ypes of Skewness
####  Positive Skewness (Right Skewed)
- Tail extends toward the right.
- Most values are relatively small.
- A few values are very large.
- Mean is generally greater than the median.
#### Examples:
- Salaries
- House prices
- Customer spending
- Hospital bills
### Negative Skewness (Left Skewed)
- Tail extends toward the left.
- Most values are relatively large.
- A few values are very small.
- Mean is generally less than the median.
#### Examples:
- Scores in an easy examination.
- Age at retirement in some selected populations.
### Symmetrical Distribution
- Both sides are approximately balanced.
- Mean and median are generally close.
#### Example:
- Some measurement errors or approximately normally distributed physical measurements.

##Log Transformation
Definition
Log transformation applies a logarithmic function to numerical values.
It is commonly used to reduce positive skewness and compress large values more strongly than small values.

#### Business Example
An e-commerce company has customer purchase values ranging from ₹100 to ₹1,00,000.

Log transformation can reduce the extreme differences between high-spending and ordinary customers, making spending patterns easier to analyze.
#### AI/ML Use Case
Log transformation is commonly useful for:
- House price prediction
- Salary prediction
- Revenue prediction
- Customer spending prediction
- Sales forecasting

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Income": [
        10000, 15000, 18000, 20000, 25000,
        30000, 40000, 50000, 100000, 500000
    ]
})

df["Log_Income"] = np.log(df["Income"])

print(df)
```

## Square Root Transformation
Definition
Square root transformation replaces each value with its square root.
It is a relatively mild transformation compared with log transformation and can reduce moderate right skewness.
It is commonly used with non-negative data, including count-based variables.
#### Mathematical Formula
\[
X_{transformed}=\sqrt{X}
\]

#### Real-World Example
Original values:
[4, 9, 16, 25, 100]
Transformed values:
[2, 3, 4, 5, 10]
The larger values are compressed.
#### Business Example
A company records the number of products purchased by customers.
Most customers buy 1–5 products, while a few customers buy 100 products.
Square root transformation can reduce the skewness of product counts.
#### AI/ML Use Case
Useful for:
- Product purchase counts
- Website clicks
- Number of transactions
- Event counts
- Moderately skewed numerical features

```python
df = pd.DataFrame({
    "Orders": [1, 2, 3, 4, 5, 8, 10, 20, 50, 100]
})

df["Sqrt_Orders"] = np.sqrt(df["Orders"])

print(df)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df["Orders"], kde=True)
plt.title("Before Square Root")

plt.subplot(1, 2, 2)
sns.histplot(df["Sqrt_Orders"], kde=True)
plt.title("After Square Root")

plt.show()
```

## Power Transformation
Definition
Power transformation applies a mathematical power-based transformation to numerical data.
It can help reduce skewness, stabilize variance, and make data more suitable for statistical modeling.
Different power transformations use different mathematical formulas and have different restrictions.
Common Power Transformations
- Square root transformation: power = 0.5
- Cube root transformation: power = 1/3
- Squaring transformation: power = 2
#### Real-World Example
Original data:
[1, 8, 27, 64, 125]
Cube root transformation:
[1, 2, 3, 4, 5]
#### Business Example
A business analyzes product sales counts that are heavily concentrated around small values.
A suitable power transformation may help reduce skewness and improve the behavior of statistical models.

```python
df = pd.DataFrame({
    "Sales": [1, 8, 27, 64, 125, 216, 343]
})

df["Cube_Root_Sales"] = np.cbrt(df["Sales"])

print(df)
```

## Box-Cox Transformation
Definition
Box-Cox is a power transformation technique used to reduce skewness and make numerical data more symmetrical.
Instead of manually choosing a power such as square root or cube root, Box-Cox estimates an appropriate transformation parameter called lambda (λ).
Important: Box-Cox requires all input values to be strictly positive.

#### Business Example
A bank analyzes customer income to build statistical models for:
- Loan risk analysis
- Income prediction
- Customer segmentation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import boxcox

df = pd.DataFrame({
    "Income": [
        10000, 12000, 15000, 20000,
        30000, 50000, 100000, 200000
    ]
})

df["BoxCox_Income"], lambda_value = boxcox(df["Income"])

print("Lambda:", lambda_value)
print(df)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df["Income"], kde=True)
plt.title("Before Box-Cox")

plt.subplot(1, 2, 2)
sns.histplot(df["BoxCox_Income"], kde=True)
plt.title("After Box-Cox")

plt.show()
```

## Yeo-Johnson Transformation
Definition
Yeo-Johnson is a power transformation technique similar to Box-Cox.
It automatically estimates a suitable lambda value to reduce skewness and stabilize variance.
Its major advantage is that it can handle positive, zero, and negative values.

#### Business Example
A company analyzes profit and loss:
[-50000, -10000, 0, 10000, 20000, 100000]
Negative values represent losses, and positive values represent profits.
Yeo-Johnson can transform this feature while preserving the ability to represent both losses and profits.
#### AI/ML Use Case
Useful for:
- Profit prediction
- Financial analysis
- Performance differences
- Features containing zero and negative values
- Regression models

## Handling Skewed Data
Definition
Handling skewed data means identifying and treating numerical features whose distributions are asymmetric.
The objective is to make the data more suitable for analysis or modeling when the model benefits from reduced skewness.
Not every skewed feature must be transformed. The choice depends on the algorithm, business meaning, and data distribution.
#### Real-World Example
Suppose house prices are:
[10, 12, 15, 18, 20, 25, 30, 100, 200]
Most houses are priced relatively low, while a few houses have extremely high prices.
This creates positive skewness.
#### Business Example
An e-commerce company analyzes customer revenue.
Most customers generate low revenue, but a few premium customers generate very high revenue.
The company may transform revenue to improve statistical analysis or regression modeling.

## Normalization
Definition
Normalization is a technique that scales numerical values into a specified range.

Min-Max normalization commonly transforms values into the range 0 to 1.

It changes the numerical scale of the feature but does not necessarily make its distribution normally distributed.

#### Business Example
An online shopping company has features such as:
- Customer age: 18–70
- Annual income: ₹1 lakh–₹50 lakh
- Number of purchases: 1–100
Normalization can bring these features onto comparable numerical ranges for algorithms that are sensitive to scale.

#### Important Note
Normalization and log transformation are different:
- Normalization changes the numerical range.
- Log transformation changes the mathematical shape of the values and can reduce skewness.
Min-Max normalization does not automatically remove skewness.

## Standardization
Definition
Standardization converts numerical values into a representation with mean approximately 0 and standard deviation approximately 1.
It is also called Z-score scaling.
It changes the scale of the feature without forcing values into a fixed range.

## Business Example
A company wants to compare employee performance across different departments.
- Department A scores: 10–100
- Department B scores: 100–1000
Standardization helps express each observation relative to its feature's mean and standard deviation.
