# Duplicate Data Handling

Duplicate Data Handling is an important step in Data Preprocessing. It helps identify repeated records, remove unnecessary duplicates, and maintain data quality before performing analysis or building Machine Learning models.

## What are Duplicate Records?
Definition
- Duplicate records are rows that contain the same or repeated information in a dataset.
- They may occur because of repeated data entry, file loading, system errors, or repeated transactions.
- Duplicate records can affect data accuracy, statistical analysis, and Machine Learning model performance.

#### Business example
A company accidentally imports the same customer file twice into its CRM system. As a result, the customer count becomes incorrect, and the sales team may contact the same customer multiple times.
#### AI/ML use case
Duplicate records in a training dataset may cause certain observations to receive more importance than others.
#### For example:
- A duplicated fraud transaction may influence model training repeatedly.
- Duplicate customer records can create incorrect customer behavior analysis.
- Repeated images in an image dataset may cause a model to memorize certain examples.

```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 101],
    "Name": ["Rahul", "Priya", "Rahul"],
    "City": ["Hyderabad", "Chennai", "Hyderabad"]
}

df = pd.DataFrame(data)

print(df)
```

#### Explanation

## Exact Duplicates
Definition
- Exact duplicates are records where all column values are identical.
- Two rows are considered exact duplicates when every corresponding value matches.
- Exact duplicates are generally easier to identify and handle.

#### Business example
A payment system processes the same payment confirmation twice and stores identical transaction records.
#### AI/ML use case
Removing exact duplicates from a training dataset helps avoid unnecessary repetition and reduces the possibility of overrepresenting identical samples.

```python
import pandas as pd

df = pd.DataFrame({
    "ID": [1, 2, 1],
    "Name": ["Ravi", "Sneha", "Ravi"],
    "Age": [25, 24, 25]
})

print(df.duplicated())
```

#### Explanation
- The first row is not considered a duplicate because no earlier row matches it.
- The second row is also unique.
- The third row exactly matches the first row.
- Pandas marks the repeated row as True.

## Partial Duplicates
Definition
- Partial duplicates are records that match on some columns but differ in other columns.
- They may represent the same entity with inconsistent or updated information.
- Identifying partial duplicates requires selecting important columns or business identifiers.

#### Business example
A CRM system stores the same customer twice because the customer updated their address.
#### AI/ML use case
Partial duplicates can affect:
- Customer segmentation.
- Customer churn prediction.
- Sales forecasting.
- Customer lifetime value calculations.

```python
df = pd.DataFrame({
    "Customer_ID": [101, 101, 102],
    "Name": ["Rahul", "Rahul", "Priya"],
    "City": ["Hyderabad", "Bengaluru", "Chennai"]
})

print(df.duplicated(subset=["Customer_ID"]))
```

#### Explanation
- We checked duplicates using only the Customer_ID column.
- Customer 101 appears twice.
- Pandas marks the second occurrence as a duplicate based on Customer_ID.
- It does not consider the City column when performing this check.

## Identifying Duplicates
Definition
- Identifying duplicates means finding repeated records in a dataset.
- Duplicate detection can be performed using all columns or selected columns.
- This step helps understand how many duplicate records exist before deciding how to handle them.
#### Real-world example
A company wants to check whether its employee dataset contains repeated employee records.
#### Business example
An HR department checks employee IDs before generating salary reports to avoid processing the same employee record multiple times.
#### AI/ML use case
Duplicate detection is used during dataset cleaning to improve data consistency and prevent duplicated samples from influencing model evaluation.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, 101, 103],
    "Name": ["Amit", "Ravi", "Amit", "Sneha"],
    "Department": ["IT", "HR", "IT", "Finance"]
})

# Identify duplicate rows
print(df.duplicated())

# Display duplicate records
print(df[df.duplicated()])

# Count duplicate records
print(df.duplicated().sum())
```

#### Explanation
- duplicated() identifies duplicate rows.
- df[df.duplicated()] displays only the duplicate records.
- duplicated().sum() counts how many duplicate records were identified.
- By default, the first occurrence is treated as the original record.

## duplicated()
Definition
- duplicated() is a Pandas method used to identify duplicate rows.
- It returns Boolean values (True or False) indicating whether a row is duplicated.
- By default, it marks later occurrences as duplicates and keeps the first occurrence unmarked.

```python
import pandas as pd

df = pd.DataFrame({
    "ID": [101, 102, 101, 101],
    "Name": ["Rahul", "Priya", "Rahul", "Rahul"]
})

# Default behavior
print(df.duplicated())

# Keep last occurrence
print(df.duplicated(keep="last"))

# Mark all duplicate occurrences
print(df.duplicated(keep=False))
```

#### Explanation
- keep="first" marks repeated rows except the first.
- keep="last" marks repeated rows except the last.
- keep=False marks every row that belongs to a duplicate group.
- These options are useful when inspecting duplicate groups before removal.

## drop_duplicates()
Definition
- drop_duplicates() removes duplicate rows from a DataFrame.
- It can remove duplicates based on all columns or selected columns.
- By default, it keeps the first occurrence and returns a cleaned DataFrame.
#### Real-world example
A dataset contains the same employee record multiple times. We want to retain only one copy.
#### Business example
A company receives the same customer master file multiple times and wants to remove repeated customer records before loading the data into its database.
#### AI/ML use case
Removing unnecessary duplicate samples can help ensure the training dataset contains unique observations.

```python
import pandas as pd

df = pd.DataFrame({
    "ID": [101, 102, 101, 103],
    "Name": ["Rahul", "Priya", "Rahul", "Amit"],
    "Age": [25, 24, 25, 26]
})

# Remove exact duplicate rows
clean_df = df.drop_duplicates()

print(clean_df)
```

#### Explanation
- drop_duplicates() compares all columns by default.
- The repeated Rahul record is removed.
- The first occurrence is retained.
- The original DataFrame remains unchanged unless we assign the result back or use inplace=True.

```python
# Keep the last duplicate
df.drop_duplicates(keep="last")

# Remove all rows belonging to duplicate groups
df.drop_duplicates(keep=False)

# Remove duplicates using selected columns
df.drop_duplicates(subset=["ID"])

# Remove duplicates using ID and Name
df.drop_duplicates(subset=["ID", "Name"])
```

## Duplicate Detection Based on Selected Columns
Definition
- Sometimes records differ in certain fields but represent the same entity.
- Selected-column duplicate detection checks repeated values only in specific columns.
- This is useful when a business identifier uniquely represents a record.

#### Business example
A CRM system uses email addresses to identify duplicate customer profiles.
#### AI/ML use case
Before building a customer segmentation model, duplicate customers can be detected using email, phone number, or customer ID.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Rahul", "Rahul Reddy", "Priya"],
    "Email": ["rahul@gmail.com",
              "rahul@gmail.com",
              "priya@gmail.com"]
})

# Identify duplicate emails
print(df[df.duplicated(subset=["Email"], keep=False)])

# Remove duplicate emails
clean_df = df.drop_duplicates(subset=["Email"], keep="first")

print(clean_df)
```

#### Explanation
- subset=["Email"] checks duplicates only using the Email column.
- keep=False identifies all records sharing a duplicate email.
- drop_duplicates() keeps only the first record for each email.
- However, the second customer record might contain updated or valuable information.

## Handling Duplicate Records
Definition
- Handling duplicates means deciding what to do with repeated records after identifying them.
- Different duplicates may require different treatment depending on their meaning.
- The correct approach depends on data quality requirements and business rules.

#### Real-world example
A customer appears twice with different phone numbers. Instead of deleting one row, both phone numbers may need to be merged into a single customer profile.
#### Business example
A company identifies duplicate supplier records. It merges supplier details, retains the valid payment information, and preserves transaction history.
#### AI/ML use case
Duplicate records can be:
- Removed from image or text datasets.
- Merged in customer datasets.
- Aggregated in repeated measurement datasets.
- Flagged for manual review in fraud detection datasets.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 102],
    "Name": ["Rahul", "Rahul", "Priya"],
    "Phone": ["9876543210", "9123456780", "9988776655"]
})

# Identify repeated customer IDs
duplicates = df[df.duplicated(
    subset=["Customer_ID"], keep=False
)]

print("Duplicate records:")
print(duplicates)

# Flag records for review
df["Duplicate_Flag"] = df.duplicated(
    subset=["Customer_ID"], keep=False
)

print(df)
```

#### Explanation
- We identified repeated Customer_ID values.
- We displayed the repeated customer records.
- We created a Duplicate_Flag column.
- Instead of immediately deleting records, we marked them for review.

## Business Rules for Duplicate Removal
Definition
- Business rules define how duplicate records should be identified and handled.
- They specify which fields uniquely identify a record and which records should be retained.
- Business rules help prevent the accidental removal of valid data.
#### Real-world example
Two students have the same name and age but different student IDs. They are different students and should not be treated as duplicates.
#### Business example
A company may define these rules:
- Customer ID must be unique.
- Email should be unique for customer profiles.
- Transaction ID must be unique for payment transactions.
- The latest verified customer address should be retained.
- Completed transactions must not be deleted merely because their amounts are identical.
#### AI/ML use case
Business rules are important when cleaning:
- Customer datasets.
- Fraud transaction datasets.
- Medical records.
- Sales and revenue datasets.
- Employee datasets.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 102],
    "Name": ["Rahul", "Rahul", "Priya"],
    "Updated_Date": [
        "2026-10-01",
        "2026-11-01",
        "2026-10-15"
    ],
    "City": ["Hyderabad", "Bengaluru", "Chennai"]
})

df["Updated_Date"] = pd.to_datetime(df["Updated_Date"])

# Business rule:
# Keep the most recently updated customer record
df = df.sort_values("Updated_Date")

clean_df = df.drop_duplicates(
    subset=["Customer_ID"],
    keep="last"
)

print(clean_df)
```

#### Explanation
- We converted the update date into a datetime format.
- We sorted the records by update date.
- We identified duplicate customers using Customer_ID.
- We retained the latest updated record for each customer.

## Why blindly deleting duplicates can be dangerous
Definition
Blindly deleting duplicates means removing repeated-looking records without understanding their meaning or checking business requirements.

A duplicate-looking record is not always an incorrect record.
