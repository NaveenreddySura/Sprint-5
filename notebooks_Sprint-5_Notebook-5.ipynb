# Data Validation

Data Validation is an important step in Data Preprocessing. It helps ensure that the dataset contains correct, meaningful, consistent, and acceptable values before performing Data Analysis or building Machine Learning models.

## What is Data Validation?
Definition
- Data Validation is the process of checking whether data satisfies predefined conditions or rules.
- It ensures that the data is accurate, consistent, and acceptable for analysis.
- Invalid records can be corrected, rejected, flagged, or sent for further investigation.
#### Real-world example
Suppose a registration form asks for age.
If a user enters:
Age = -25
The system should reject the value because a person's age cannot be negative.
#### Business example
A company receives employee data containing salary, joining date, department, and employee ID.

#### Data Validation checks whether:
- Salary is positive or zero.
- Department belongs to an approved list.
- Joining date is valid.
- Employee ID is not missing.
#### AI/ML use case
Data Validation ensures that training datasets contain valid values.
For example:
- Negative ages can distort customer analysis.
- Invalid salary values can affect salary prediction.
- Incorrect categories can create unexpected values during encoding.
- Invalid dates can affect time-based predictions.

## Range Validation
Definition
- Range Validation checks whether a numerical value falls within an acceptable minimum and maximum range.
- It is used for values such as age, salary, percentage, marks, and ratings.
- Values outside the permitted range are identified as invalid.
#### Real-world example
A student's percentage must be between 0 and 100.
Percentage = 105
This is invalid because the maximum percentage is 100.
#### Business example
A company provides a performance rating from 1 to 5.

Any employee rating outside this range must be reviewed.
#### AI/ML use case
Range Validation prevents incorrect numerical values from entering Machine Learning datasets.
For example:
- Age cannot be negative.
- Customer satisfaction score must be within its defined scale.
- Discount percentage must be within an approved range.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [25, -5, 150],
    "Percentage": [85, 105, 60]
})

# Age validation: 0 to 120
valid_age = df["Age"].between(0, 120)

# Percentage validation: 0 to 100
valid_percentage = df["Percentage"].between(0, 100)

print("Valid Age:")
print(valid_age)

print("\nValid Percentage:")
print(valid_percentage)

# Display invalid records
print("\nInvalid Age Records:")
print(df[~valid_age])

print("\nInvalid Percentage Records:")
print(df[~valid_percentage])
```

#### Explanation
- between(0, 120) checks whether age falls between 0 and 120.
- between(0, 100) checks whether percentage falls between 0 and 100.
- ~ reverses the Boolean condition.
- df[~valid_age] displays records with invalid age values.

## Type Validation
Definition
- Type Validation checks whether a column contains the expected data type.
- It ensures numerical fields contain numerical values, date fields contain dates, and categorical fields contain appropriate text or category values.
- Incorrect data types can cause calculation errors and unexpected results.
#### Real-world example
A form expects age as a number.
Instead of:
Age = 25
Someone enters:
Age = "Twenty Five"
The value cannot be directly used in numerical calculations.
#### Business example
An employee dataset expects salary to be numeric. If salary is stored as text, payroll calculations may fail.
#### AI/ML use case
Machine Learning algorithms generally require numerical input for numerical features and appropriate encoding for categorical features.
Incorrect data types can cause:
- Training errors.
- Incorrect calculations.
- Failed preprocessing pipelines.
- Unexpected model behavior.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": ["25", "30", "Twenty"],
    "Salary": ["50000", "40000", "35000"]
})

print(df.dtypes)

# Convert Age to numeric
df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

# Convert Salary to numeric
df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

print("\nAfter Conversion:")
print(df)

print("\nUpdated Data Types:")
print(df.dtypes)
```

#### Explanation
- Initially, Age and Salary are stored as object/string columns.
- pd.to_numeric() converts valid numerical strings into numbers.
- errors="coerce" converts invalid values into NaN.
- "Twenty" becomes NaN because it cannot be converted into a number.

## Format Validation
Definition
- Format Validation checks whether a value follows a predefined structure or pattern.
- It is commonly used for email addresses, phone numbers, postal codes, and identification numbers.
- A value may contain text but still be invalid if its format is incorrect.
#### Real-world example
An email address should follow a structure similar to:
username@example.com
An email such as:
rahulgmail.com
does not contain the required @ symbol.
#### Business example
A company validates customer email addresses and phone numbers before storing them in a CRM system.
#### AI/ML use case
Format Validation helps clean structured customer information before performing customer segmentation, marketing analysis, or fraud detection.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Email": [
        "rahul@gmail.com",
        "priyagmail.com",
        "amit@yahoo.com"
    ]
})

# Basic email pattern check
email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

df["Email_Valid"] = df["Email"].str.match(
    email_pattern,
    na=False
)

print(df)

print("\nInvalid Emails:")
print(df[~df["Email_Valid"]])
```

#### Explanation
- str.match() checks whether each email follows the specified pattern.
- na=False treats missing email values as not matching the pattern.
- Email_Valid stores True for matching emails and False for invalid ones.
- The regular expression performs a basic structural check, not full verification that an email address exists.

## Category Validation
Definition
- Category Validation checks whether categorical values belong to an approved set of values.
- It is used for columns such as gender, department, product category, payment method, and status.
- Unexpected category values are identified as invalid.
#### Real-world example
A form accepts these payment methods:
UPI
Cash
Card
If someone enters:
Payment = Bitcoin
It is invalid if Bitcoin is not an accepted payment method in that system.
#### Business example
A company accepts employee departments such as:
- IT
- HR
- Finance
- Sales
If a record contains Marketingg, it may be a spelling mistake or an unapproved department.
#### AI/ML use case
Category Validation prevents unexpected categories from entering a Machine Learning pipeline.

For example, a model trained on IT, HR, and Finance may encounter an unexpected category during prediction.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Department": ["IT", "HR", "Finance", "Marketingg"]
})

# Accepted categories
allowed_departments = ["IT", "HR", "Finance", "Sales"]

# Validate categories
df["Department_Valid"] = df["Department"].isin(
    allowed_departments
)

print(df)

print("\nInvalid Categories:")
print(df[~df["Department_Valid"]])
```

#### Explanation
- isin() checks whether each department belongs to the accepted list.
- Valid departments return True.
- Marketingg returns False.
- Invalid categories can be corrected, mapped, or flagged for review.

## Null Validation
Definition
- Null Validation checks whether required fields contain missing values.
- It ensures that important columns are not empty or missing.
- Missing values in mandatory fields may cause calculations, processing, or business operations to fail.
#### Real-world example
A registration form requires a customer name.

If the customer name is missing, the registration should not be completed.
#### Business example
A company requires employee ID, employee name, and joining date for every employee record.

If employee ID is missing, the record cannot be uniquely identified.
#### AI/ML use case
Missing values in features may cause model training errors or reduce prediction quality.

Null Validation identifies missing values before applying imputation or other preprocessing methods.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, None],
    "Name": ["Rahul", None, "Amit"],
    "Salary": [50000, 40000, 35000]
})

# Check missing values
print(df.isnull())

# Validate mandatory columns
required_columns = ["Employee_ID", "Name"]

df["Required_Fields_Valid"] = (
    df[required_columns].notnull().all(axis=1)
)

print("\nValidation Result:")
print(df)

print("\nInvalid Records:")
print(df[~df["Required_Fields_Valid"]])
```

#### Explanation
- isnull() identifies missing values.
- notnull() identifies values that are present.
- all(axis=1) checks whether all required columns are valid for each row.
- Records with missing Employee_ID or Name are marked invalid.

## Business Rule Validation
Definition
- Business Rule Validation checks whether data satisfies specific business requirements.
- These rules may involve one or more columns.
- Business rules are defined according to the company's operations, policies, and domain requirements.
#### Real-world example
A person cannot have an age below zero.

A person who is 10 years old cannot normally have 15 years of work experience.
#### Business example
An employee's joining date should not be earlier than their date of birth.

A company may also require that:
- Salary should not be negative.
- Discount should not exceed the approved limit.
- Order quantity should be greater than zero.
- Employee experience should not exceed the possible working age.
#### AI/ML use case
Business Rule Validation prevents inconsistent records from affecting Machine Learning analysis.

For example, invalid salary or experience values can distort salary prediction models.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [25, 20, 30],
    "Experience": [3, 25, 5],
    "Salary": [50000, -1000, 60000]
})

# Rule 1: Age must be non-negative
rule_age = df["Age"] >= 0

# Rule 2: Experience cannot exceed age
rule_experience = df["Experience"] <= df["Age"]

# Rule 3: Salary must be non-negative
rule_salary = df["Salary"] >= 0

# All business rules must pass
df["Business_Valid"] = (
    rule_age & rule_experience & rule_salary
)

print(df)

print("\nInvalid Records:")
print(df[~df["Business_Valid"]])
```

#### Explanation
- rule_age checks whether age is non-negative.
- rule_experience checks whether experience is less than or equal to age.
- rule_salary checks whether salary is non-negative.
- & combines the conditions using AND.
- A record is valid only when all three conditions are satisfied.

## Referential Validation
Definition
- Referential Validation checks whether a value in one dataset correctly references a record in another dataset.
- It is commonly used to validate foreign keys and relationships between tables.
- It ensures that related records exist in the appropriate master dataset.
#### Real-world example
An order contains a Customer_ID.

That Customer_ID must exist in the customer table.
If an order references Customer_ID 999 but no such customer exists, the reference is invalid.
#### Business example
A company's order table references customers from the customer master table.

Every Order.Customer_ID must match a valid Customer.Customer_ID.
#### AI/ML use case
Referential Validation ensures that customer, product, and transaction datasets are correctly linked before performing customer analysis or building predictive models.

```python
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Rahul", "Priya", "Amit"]
})

orders = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003],
    "Customer_ID": [101, 102, 999]
})

# Check whether order customer IDs exist
orders["Customer_Valid"] = orders["Customer_ID"].isin(
    customers["Customer_ID"]
)

print(orders)

print("\nInvalid Customer References:")
print(orders[~orders["Customer_Valid"]])
```

#### Explanation
- The customer table contains IDs 101, 102, and 103.
- The order table contains customer IDs 101, 102, and 999.
- isin() checks whether the order customer ID exists in the customer master.
- Customer_ID 999 is invalid because it does not exist in the customer table.

## Date Validation
Definition
- Date Validation checks whether date values are valid and follow logical date rules.
- It ensures that dates are correctly formatted and chronologically consistent.
- It is useful for employee records, financial transactions, orders, and time-based datasets.
#### Real-world example
A person's joining date cannot be before their date of birth.
#### Business example
An employee joining date should be after their date of birth.

An order delivery date should not be before the order date.
#### AI/ML use case
Date Validation is important for:
- Time series forecasting.
- Employee attrition prediction.
- Customer churn prediction.
- Sales forecasting.
- Financial analysis.
Incorrect dates can cause errors in calculating durations and trends

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "DOB": ["2000-01-10", "2002-05-15", "1998-08-20"],
    "Joining_Date": [
        "2022-06-01",
        "2000-01-01",
        "2020-03-10"
    ]
})

# Convert date columns
df["DOB"] = pd.to_datetime(
    df["DOB"],
    errors="coerce"
)

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)

# Validate date sequence
df["Date_Valid"] = df["Joining_Date"] >= df["DOB"]

print(df)

print("\nInvalid Date Records:")
print(df[~df["Date_Valid"]])
```

#### Explanation
- pd.to_datetime() converts string values into datetime values.
- errors="coerce" converts invalid date strings into NaT.
- The comparison checks whether joining date is on or after date of birth.
- Records with an invalid date sequence are marked False.

## Constraint Validation
Definition
- Constraint Validation checks whether data satisfies predefined restrictions or constraints.
- Constraints may apply to individual columns or relationships between multiple columns.
- They are commonly used to enforce uniqueness, required values, valid ranges, and relationships between records.
#### Real-world example
A student ID must be unique.

Two students cannot have the same student ID in the same student database.
#### Business example
A company may enforce:
- Employee_ID must be unique.
- Customer_ID must not be null.
- Salary must be non-negative.
- Order quantity must be positive.
- Invoice amount must match the calculated total.
#### AI/ML use case
Constraint Validation helps maintain a consistent dataset before training models.

For example:
- Unique customer identifiers prevent accidental record duplication.
- Positive quantities prevent incorrect sales calculations.
- Valid feature ranges reduce data quality problems.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, 101, 104],
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Salary": [50000, 40000, 60000, -5000]
})

# Constraint 1: Employee ID must be unique
unique_id = ~df["Employee_ID"].duplicated(keep=False)

# Constraint 2: Salary must be non-negative
valid_salary = df["Salary"] >= 0

# Constraint 3: Employee name must not be null
valid_name = df["Name"].notnull()

# Combine constraints
df["Constraint_Valid"] = (
    unique_id & valid_salary & valid_name
)

print(df)

print("\nInvalid Records:")
print(df[~df["Constraint_Valid"]])
```

#### Explanation
- duplicated(keep=False) identifies every occurrence of a duplicate Employee_ID.
- ~ reverses the result, so unique IDs become True.
- Salary >= 0 checks the salary constraint.
- notnull() checks whether employee name exists.
- All constraints are combined using.
