# Encoding Categorical Variables
Categorical Encoding is an important step in Data Preprocessing. It converts categorical information into a numerical representation that Machine Learning algorithms can understand.

Many Machine Learning algorithms work with numerical inputs. Therefore, categorical variables such as gender, department, education level, product category, and payment method often need to be encoded before training a model.

## What is Categorical Encoding?
Definition
- Categorical Encoding is the process of converting categorical variables into numerical representations.
- It allows Machine Learning algorithms to process categorical information.
- The encoding technique depends on whether the categories are nominal, ordinal, binary, or high-cardinality.

#### Business example
An e-commerce company wants to predict whether customers will purchase a product.
Its dataset contains:
- Gender
- Product category
- Payment method
- Membership level
These categorical variables must be appropriately encoded before model training.
#### AI/ML use case
Categorical Encoding is used in:
- Customer churn prediction.
- Fraud detection.
- Recommendation systems.
- House price prediction.
- Employee attrition prediction.
- Customer segmentation.

## Types of Categorical Variables
Before choosing an encoding technique, understand the type of categorical variable.
#### Nominal Variables
Definition
- Nominal variables contain categories without any natural order.
- One category is not greater or smaller than another.
- Assigning arbitrary numerical ranks can create misleading relationships.

#### Business example
A company wants to predict employee attrition using Department.
IT, HR, and Finance do not have any natural ranking.
#### AI/ML use case
Nominal variables are commonly encoded using:
- One-Hot Encoding.
- Dummy Encoding.
- Frequency Encoding.
- Target Encoding for suitable high-cardinality features.

```python
#### import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "IT"]
})

encoded = pd.get_dummies(
    df,
    columns=["Department"],
    dtype=int
)

print(encoded)
```

## Ordinal Variables
Definition
- Ordinal variables contain categories with a meaningful natural order.
- Categories can be ranked from lower to higher.
- The difference between consecutive categories may not be equal.

#### AI/ML use case
Ordinal Encoding is useful for:
- Customer satisfaction prediction.
- Credit risk classification.
- Education-based analysis.
- Product quality ratings.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Satisfaction": [
        "Low", "Medium", "High", "Low"
    ]
})

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["Satisfaction_Encoded"] = encoder.fit_transform(
    df[["Satisfaction"]]
)

print(df)
```

## Label Encoding
Definition
- Label Encoding assigns a numerical integer to each categorical value.
- Each category receives a unique integer.
- It is generally appropriate for target labels or certain tree-based modeling workflows, but can introduce artificial ordering for nominal input features.
#### Real-world example
A dataset contains:
Red
Blue
Green
Label Encoding may convert them into:
Red = 0
Blue = 1
Green = 2
#### Business example
A company classifies customer feedback into:
Positive
Negative
Neutral
These categories can be encoded as numerical class labels for classification.
#### AI/ML use case
Label Encoding is commonly used for:
- Classification target variables.
- Machine Learning class labels.
- Certain tree-based workflows where categorical handling is supported appropriately.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Feedback": [
        "Positive", "Negative",
        "Neutral", "Positive"
    ]
})

encoder = LabelEncoder()

df["Feedback_Encoded"] = encoder.fit_transform(
    df["Feedback"]
)

print(df)
```

#### Explanation
- LabelEncoder() creates an encoder.
- fit_transform() learns the categories and converts them into integers.
- The mapping is determined by the encoder's category ordering.
#### When to use Label Encoding
- For target labels such as Yes/No or class names.
- When the algorithm or workflow appropriately handles integer-coded categories.
- For certain tree-based workflows where the category mapping does not imply a misleading numeric relationship.
#### When not to use Label Encoding
For nominal input features with models such as:
- Linear Regression.
- Logistic Regression.
- K-Nearest Neighbors.
- Distance-based clustering.

## Ordinal Encoding
Definition
- Ordinal Encoding converts categories into numbers according to their meaningful natural order.
- Each category receives a numerical rank.
- The order must be explicitly defined according to the business meaning of the variable.

#### AI/ML use case
Ordinal Encoding is useful for:
- Satisfaction prediction.
- Credit risk analysis.
- Education-level analysis.
- Product quality ratings.
- Customer loyalty levels.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Risk": [
        "Low", "High", "Medium", "Low"
    ]
})

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["Risk_Encoded"] = encoder.fit_transform(
    df[["Risk"]]
)

print(df)
```

#### Explanation
- We explicitly defined the order.
- Low Risk becomes 0.
- Medium Risk becomes 1.
- High Risk becomes 2.
- The encoded values represent the increasing risk level.

## One-Hot Encoding
Definition
- One-Hot Encoding converts each category into a separate binary column.
- Each observation receives 1 in the column corresponding to its category and 0 in the other columns.
- It avoids introducing artificial numerical ordering between nominal categories.

#### Business example
An e-commerce company wants to predict customer purchases based on product category.

Categories:
Electronics
Clothing
Furniture
Each category becomes a separate binary feature.
#### AI/ML use case
One-Hot Encoding is useful for:
- Logistic Regression.
- Linear Regression.
- Neural networks.
- KNN.
- Customer churn prediction.
- House price prediction.

```python
import pandas as pd

df = pd.DataFrame({
    "City": [
        "Hyderabad",
        "Chennai",
        "Bengaluru",
        "Hyderabad"
    ]
})

encoded = pd.get_dummies(
    df,
    columns=["City"],
    dtype=int
)

print(encoded)
```

#### Explanation
- pd.get_dummies() converts categories into binary columns.
- Every row receives 1 for its corresponding category.
- Other category columns receive 0.
- The encoding does not assign numerical rankings to cities.

## Dummy Variables
Definition
- Dummy Encoding represents categorical variables using binary columns.
- For a categorical feature with \(k\) categories, it generally creates \(k-1\) columns when one category is dropped.
- The omitted category becomes the reference or baseline category.

#### Business example
A company wants to understand how employee department affects salary.
Finance is chosen as the baseline department.

The model compares IT and HR against Finance.
#### AI/ML use case
Dummy Encoding is useful for:
- Linear Regression.
- Logistic Regression.
- Statistical modeling.
- Regression with categorical predictors.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": [
        "IT", "HR", "Finance", "IT"
    ]
})

encoded = pd.get_dummies(
    df,
    columns=["Department"],
    drop_first=True,
    dtype=int
)

print(encoded)
```

#### Explanation
- drop_first=True removes one category column.
- Finance becomes the reference category.
- HR and IT are represented relative to Finance.
- This can avoid perfect multicollinearity caused by including all dummy columns alongside an intercept in linear models.

## Frequency Encoding
Definition
- Frequency Encoding replaces each category with the number of times it appears in the dataset.
- Instead of creating multiple columns, it creates one numerical column.
- It is useful for categorical variables with many unique categories.
#### Real-world example
Suppose a dataset contains:
City
Hyderabad
Chennai
Hyderabad
Bengaluru
Hyderabad
Chennai
Frequencies:
Hyderabad = 3
Chennai = 2
Bengaluru = 1
Encoded values:
Hyderabad = 3
Chennai = 2
Bengaluru = 1


#### Business example
An e-commerce company has thousands of product brands.
Instead of creating thousands of one-hot columns, it encodes each brand based on its occurrence frequency.
#### AI/ML use case
Frequency Encoding is useful for:
- High-cardinality categorical features.
- Product brands.
- City names.
- Customer segments.
- Supplier names.
- Large categorical datasets.

```python
import pandas as pd

df = pd.DataFrame({
    "City": [
        "Hyderabad",
        "Chennai",
        "Hyderabad",
        "Bengaluru",
        "Hyderabad",
        "Chennai"
    ]
})

# Calculate frequency
frequency = df["City"].value_counts()

# Encode categories
df["City_Frequency"] = df["City"].map(
    frequency
)

print(df)
```

#### Explanation
- value_counts() calculates category frequencies.
- map() replaces each city with its frequency.
- Hyderabad becomes 3.
- Chennai becomes 2.
- Bengaluru becomes 1.

## Target Encoding
Definition
- Target Encoding replaces a categorical value with a statistic calculated from the target variable.
- For regression, this is commonly the mean target value for each category.
- For classification, it may use the category's positive-class probability or another target-related statistic.

#### Business example
An e-commerce company wants to predict whether customers will purchase a product.
The city category may be encoded using the average purchase rate observed for that city.
#### AI/ML use case
Target Encoding is useful for:
- House price prediction.
- Customer purchase prediction.
- Customer churn prediction.
- Fraud prediction.
- High-cardinality categorical features.

```python
import pandas as pd

df = pd.DataFrame({
    "Locality": ["A", "A", "B", "B", "C"],
    "Price": [50, 60, 30, 40, 100]
})

# Calculate mean target per category
target_mean = df.groupby(
    "Locality"
)["Price"].mean()

# Encode category
df["Locality_Encoded"] = df["Locality"].map(
    target_mean
)

print(df)
```

#### Explanation
- groupby() groups records by locality.
- mean() calculates average house price for each locality.
- map() replaces locality categories with their corresponding average price.
- Locality A becomes 55.
- Locality B becomes 35.
- Locality C becomes 100.

## Handling Unknown Categories
Definition
- Unknown categories are values encountered during prediction that were not present during encoder training.
- These categories can cause errors when using encoders that expect previously observed categories.
- Unknown values must be handled using an appropriate strategy.

#### Business example
An e-commerce model is trained using product categories:
Electronics
Clothing
Furniture
A new product category called Sports appears after deployment.
The model must handle this new category without crashing.
#### AI/ML use case
Unknown category handling is important for:
- Production Machine Learning models.
- Customer churn prediction.
- Recommendation systems.
- Real-time prediction APIs.
- Models receiving new categories over time.

```python
import pandas as pd

from sklearn.preprocessing import OneHotEncoder

train = pd.DataFrame({
    "City": ["Hyderabad", "Chennai", "Bengaluru"]
})

test = pd.DataFrame({
    "City": ["Hyderabad", "Mumbai"]
})

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

train_encoded = encoder.fit_transform(train)
test_encoded = encoder.transform(test)

print("Categories:")
print(encoder.categories_)

print("\nTest Encoding:")
print(test_encoded)
```

#### Explanation
- fit_transform() learns categories from the training dataset.
- handle_unknown="ignore" prevents errors for unknown categories.
- Mumbai is not assigned a new column during prediction.
- Its encoded representation becomes all zeros for that feature.

## High-Cardinality Categories
Definition
- High-cardinality categorical variables contain a large number of unique categories.
- They can create thousands of features when One-Hot Encoding is applied.
- High-cardinality variables require encoding techniques that control memory usage and model complexity.
#### Real-world example
A dataset contains 10,000 different product IDs.
One-Hot Encoding could create 10,000 separate columns.
#### Business example
An e-commerce company has:
- 20,000 product IDs.
- 5,000 customer locations.
- 2,000 supplier IDs.
Creating a separate feature for every category may become inefficient.
#### AI/ML use case
High-cardinality encoding is useful in:
- Recommendation systems.
- Fraud detection.
- E-commerce prediction.
- Customer analytics.
- Large-scale tabular Machine Learning.

```python
import pandas as pd

df = pd.DataFrame({
    "Product_ID": [
        "P101", "P102", "P103",
        "P101", "P104", "P105"
    ]
})

# Check cardinality
print("Unique Categories:")
print(df["Product_ID"].nunique())

# Frequency Encoding
frequency = df["Product_ID"].value_counts()

df["Product_Frequency"] = df["Product_ID"].map(
    frequency
)

print(df)
```

#### Explanation
- nunique() counts unique categories.
- value_counts() calculates their frequency.
- map() converts product IDs into frequency values.
- The original high-cardinality feature is represented using one numerical column.
