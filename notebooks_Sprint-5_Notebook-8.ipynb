# Feature Scaling
What is Feature Scaling?
Feature Scaling is a data preprocessing technique used to bring numerical features into a similar scale or range.

Different features in a dataset may have different units and magnitudes. Feature scaling prevents features with larger numerical values from unnecessarily dominating features with smaller values.

## Why Feature Scaling?
Feature scaling is important because many ML algorithms calculate distances, gradients, or mathematical relationships using numerical features.

#### Main benefits
- Prevents large-scale features from dominating distance calculations.
- Helps gradient-based algorithms converge faster.
- Improves performance of scale-sensitive algorithms.
- Makes features more comparable.
- Supports PCA and other mathematical techniques that depend on feature variance.

#### Business example
A company wants to predict customer churn using:
- Age
- Monthly income
- Number of purchases
- Customer satisfaction score
- Years of membership
Without scaling, income might dominate distance-based calculations because its values are much larger than the other features.
#### AI/ML use case
Feature scaling is particularly important for:
- Customer segmentation using K-Means
- Customer similarity using KNN
- Classification using SVM
- Logistic Regression
- Neural Networks
- Principal Component Analysis (PCA)

## Scale-Sensitive Algorithms
Definition
Scale-sensitive algorithms are ML algorithms whose results, training behavior, or predictions can change significantly when numerical features have different scales.
These algorithms are affected by feature magnitudes, distances, or optimization behavior.
### KNN (K-Nearest Neighbors)
Definition
KNN is a supervised ML algorithm that predicts the class or value of a new observation by identifying its nearest neighbors in the training dataset.
It commonly uses distance calculations to find similar observations.
Why scaling is required
KNN calculates distances between observations.
Features with larger numerical values can dominate distance calculations, making smaller-scale features less influential.
#### Real-world example
Suppose a company wants to identify customers similar to an existing customer using:
- Age
- Salary
- Purchase frequency
Salary might range from ₹20,000 to ₹1,00,000, while purchase frequency ranges from 1 to 20.
Without scaling, salary differences may dominate the distance calculation.
Scaling makes these features contribute more comparably.
#### AI/ML use case
- Customer recommendation systems
- Similar customer identification
- Medical classification
- Credit risk classification
- Product recommendation

It identifies directions that capture the greatest variance in the data.

Features with larger numerical scales may have larger variances and dominate the principal components.

Scaling ensures that differences in units do not unnecessarily dominate the analysis.

## Scale-Insensitive Algorithms
Definition
Scale-insensitive algorithms are algorithms whose predictions generally do not depend strongly on the numerical scale of individual features.

Tree-based algorithms typically make decisions using feature thresholds rather than calculating distances between observations.
### Decision Tree
Definition
A Decision Tree is a supervised ML algorithm that makes predictions using a sequence of feature-based decision rules.

It divides data into smaller groups using conditions on individual features.
Why scaling is usually unnecessary
Decision Trees split data based on feature thresholds.

Changing the scale of a feature generally changes its threshold correspondingly without changing the underlying split.
#### Real-world example
A bank predicts loan approval using:
- Salary
- Age
- Credit score

## Normalization

In many introductory ML contexts, normalization refers to Min-Max Scaling, which transforms feature values into a range such as 0 to 1.

Important: In scikit-learn, Normalizer specifically means normalizing each sample to unit norm. It is different from Min-Max Scaling.
#### Real-world example
Suppose a student dataset contains marks:
[40, 60, 80, 100]

[0.00, 0.33, 0.67, 1.00]
The original values are transformed into a range between 0 and 1.

## Min-Max Scaling
Definition
Min-Max Scaling transforms numerical values into a specified range, usually 0 to 1.
It uses the minimum and maximum values of each feature to calculate scaled values.

#### Business example
A company has product ratings from 1 to 5 and wants to transform them into a 0–1 range for a model.
AI/ML use case
- Neural Networks
- KNN
- K-Means
- SVM
- Image processing

#### When to use Min-Max Scaling
Use it when:
- A fixed range is required.
- Features have reasonably controlled distributions.
- Your model benefits from bounded inputs.
- You want features to have comparable numerical ranges.
#### When not to use Min-Max Scaling
Avoid relying on it without checking outliers.An extreme observation can compress most normal values into a very small range.

## Standardization
Definition
Standardization transforms numerical features so that they have a mean of approximately 0 and a standard deviation of approximately 1.

Unlike Min-Max Scaling, standardization does not force values into a fixed range.

#### Interpretation
- Negative value means below the mean.
- Zero means equal to the mean.
- Positive value means above the mean.
- Magnitude indicates how many standard deviations away the observation is.
#### Business example
A company compares employee salaries, performance scores, and years of experience.

Standardization allows the company to express each feature relative to its own mean and variability.
#### AI/ML use case
- Logistic Regression
- SVM
- PCA
- KNN
- K-Means
- Neural Networks

#### When to use Standardization
Use it when:
- Features have different numerical ranges.
- Your model uses gradient optimization.
- Your model uses regularization.
- Your algorithm uses distances.
- You are applying PCA.
#### When not to use Standardization
It is generally unnecessary for tree-based algorithms.

It is also not a robust solution when extreme outliers heavily distort the mean and standard deviation.

## StandardScaler
Definition
StandardScaler is a scikit-learn preprocessing class that standardizes numerical features using their training-data mean and standard deviation.
It is one of the most commonly used scaling techniques in machine learning.

## RobustScaler
Definition
RobustScaler scales numerical features using the median and interquartile range (IQR).
It is designed to reduce the influence of outliers during scaling.

#### Real-world example
Consider salary data:
[20000, 25000, 30000, 35000, 40000, 1000000]
The last value is an extreme salary.
Mean and standard deviation will be affected significantly by ₹10,00,000.
Median and IQR are much less affected.

#### When to use RobustScaler
Use it when:
- Salary datasets contain extreme values.
- Financial data contains large deviations.
- Customer spending contains genuine extreme observations.
- The mean and standard deviation do not represent the central distribution well.
Important limitation
RobustScaler does not remove outliers.
An extreme observation can still become a very large scaled value.
It only makes the scaling parameters less sensitive to outliers.

## MaxAbsScaler
Definition
MaxAbsScaler scales each feature by dividing its values by the maximum absolute value of that feature.
It preserves the sign of values and does not shift the data to zero mean.

#### When to use MaxAbsScaler
Use it when:
- Working with sparse datasets.
- Working with many zero values.
- Preserving zero is important.
- Features contain positive and negative values.
- Text data is represented using sparse matrices.
#### Important limitation
MaxAbsScaler is sensitive to outliers because it uses the maximum absolute value.

One extreme observation can compress all other values toward zero.
