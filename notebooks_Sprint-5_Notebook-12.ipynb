# Train, Validation & Test Data

What is Data Splitting?
Definition
Data Splitting is the process of dividing a dataset into separate subsets for training, validating, and testing a Machine Learning model.
Each subset has a different purpose during model development.
The main objective is to evaluate how well a model performs on data it has never seen before.
#### Real-World Example
Suppose you have 10,000 student records and want to predict examination results.
You can divide the dataset into:
- Training Dataset: 7,000 records
- Validation Dataset: 1,500 records
- Test Dataset: 1,500 records
The model learns from the training data, you use validation data to improve the model, and finally you use test data to evaluate its performance.
#### Business Example
An e-commerce company wants to predict whether customers will purchase products.
It has 50,000 customer records.
The company divides the records into different datasets to:
- Train the purchase prediction model.
- Select the best model settings.
- Evaluate the final model on unseen customer records.
#### AI/ML Use Case
Data Splitting is essential for:
- Customer Churn Prediction
- Fraud Detection
- House Price Prediction
- Loan Default Prediction
- Sales Forecasting
- Medical Diagnosis
- Recommendation Systems

## Training Dataset
Definition
The Training Dataset is the portion of data used to teach a Machine Learning model.
The model identifies patterns and relationships between input features and the target variable using this dataset.

During training, the model adjusts its internal parameters to reduce prediction errors.
#### Real-World Example
Suppose you want to predict house prices using:
- House Area
- Location
- Number of Bedrooms
- Building Age
The training dataset contains historical house records with their actual prices.
The model learns how these features relate to house prices.
#### Business Example
A bank wants to predict whether a customer will default on a loan.
The training data contains:
- Customer Income
- Credit Score
- Loan Amount
- Repayment History

The model learns patterns associated with loan defaults.

## Validation Dataset
Definition
The Validation Dataset is a separate portion of data used during model development to evaluate and improve a model.
It helps compare different algorithms, tune hyperparameters, and select the best model.

The model does not directly learn its parameters from the validation dataset.
#### Real-World Example
Suppose you are building a house price prediction model.
You train three models:
- Linear Regression
- Decision Tree
- Random Forest
You evaluate their performance using validation data and choose the most suitable model.
#### Business Example
An e-commerce company builds a customer purchase prediction model.
The company experiments with:
- Different algorithms.
- Different model settings.
- Different feature combinations.
It uses validation data to decide which model performs best.
#### Why Do We Need Validation Data?
Without validation data, you may repeatedly make decisions based on training performance.
A model that performs very well on training data may not perform well on unseen data.

Validation data helps identify this problem during development.
Important Point
Validation data is used for model selection and tuning.
If you repeatedly tune your model based on validation results, you may gradually overfit to the validation dataset too.

## Test Dataset
Definition
The Test Dataset is a separate portion of data used to evaluate the final trained model.
It provides an estimate of how well the model performs on unseen data.
It should not be used for model training, feature selection, or hyperparameter tuning.
#### Real-World Example
Suppose you trained a model to predict student examination results.
After selecting the best model using validation data, you evaluate it using test data.

This tells you how well it predicts results for students whose records were not used during model development.
#### Business Example
A bank develops a loan default prediction model.
Before deploying the model, it evaluates the final model using test records that were kept separate throughout model development.
#### Why Is Test Data Important?
- Measures generalization performance.
- Helps detect overfitting.
- Provides a final performance estimate.
- Helps assess whether the model is ready for deployment.

## Train-Test Split
Definition
Train-Test Split divides a dataset into two separate portions.
One portion is used to train the model, and the other is used to evaluate it.
It is commonly used when you do not need a separate validation dataset.
#### Real-World Example
You have 1,000 house records.
You divide them into:
- 800 training records.
- 200 testing records.
The model learns from 800 records and is evaluated on the remaining 200.

## Train-Validation-Test Split
Definition
Train-Validation-Test Split divides the dataset into three separate portions.
Training data teaches the model, validation data helps select and tune the model, and test data evaluates the final model.
It provides a structured way to develop and evaluate Machine Learning models.
#### Real-World Example
Suppose you have 10,000 customer records.
You divide them into:
- Training: 7,000 records.
- Validation: 1,500 records.
- Testing: 1,500 records.

## Random State
Definition
random_state is a parameter used to control the randomness involved in splitting data.
When you split a dataset, the records are usually selected randomly. Different executions may produce different training and testing records.
Setting a fixed random_state makes the split reproducible, so you can obtain the same split when you run the code again with the same data and settings.
#### Real-World Example
Suppose you have 1,000 customer records.
You split them into training and testing datasets.

Without a fixed random state:
- The first execution may select customer records 1, 5, 8, and 10 for testing.
- The next execution may select different customers.
With a fixed random state, the same split is reproduced.

## Stratified Sampling
Definition
Stratified Sampling is a splitting technique that maintains approximately the same class proportions in each dataset.
It is especially useful for classification problems where the target classes are imbalanced.

Without stratification, random splitting may produce training and testing datasets with very different class distributions.
#### Real-World Example
Suppose a bank has 1,000 loan records:
- 900 customers did not default.
- 100 customers defaulted.
This means:
- 90% belong to the No Default class.
- 10% belong to the Default class.
If you randomly split the data, the class proportions may differ between training and testing data.
Stratified Sampling tries to preserve the original 90:10 ratio in both datasets.

## Data Leakage
Definition
Data Leakage occurs when information that would not legitimately be available during model training or prediction is used to train or evaluate a model.
This can make model performance appear better than it really is.

The model may learn information from the test dataset or from future events that would not be available when making real predictions.
#### Real-World Example
Suppose you are developing a model to predict whether a customer will default on a loan.
Your dataset contains:
- Customer Income
- Credit Score
- Loan Amount
- Repayment History
- Loan Recovery Status
If Loan Recovery Status is only known after a customer defaults, using it to predict default would introduce target leakage.
The model would be using information from the future.
#### Business Example
An e-commerce company wants to predict whether a customer will purchase a product.
The dataset contains:
- Customer Age
- Product Views
- Previous Purchases
- Payment Confirmation
If Payment Confirmation is recorded after the purchase, using it to predict whether the customer will purchase is leakage.
Types of Data Leakage
#### --Train-Test Leakage

Information from the test dataset is used during training or preprocessing.

Example: Calculating the mean and standard deviation using the entire dataset before splitting.

#### --Target Leakage

A feature contains information that directly reveals or is created using the target.

Example: Using loan recovery status to predict whether a loan will default.

#### --Temporal Leakage

Future information is used to predict an earlier event.

Example: Using next month's sales information to predict this month's sales.

#### --Duplicate or Group Leakage

The same customer, patient, or other related entity appears in both training and testing datasets when the intended evaluation requires completely unseen entities.

Example: Records from the same patient appear in both sets, allowing the model to recognize patient-specific patterns.
#### Why Is Data Leakage Dangerous?
- Produces unrealistically high evaluation scores.
- Hides poor generalization.
- Creates misleading model comparisons.
- Causes problems after deployment.
- Makes the model unreliable on real-world data.

## Temporal Splitting
Definition
Temporal Splitting is a data splitting technique used when observations are ordered by time.
Instead of randomly dividing the records, the dataset is split according to chronological order.
Older records are used for training, more recent records for validation, and the latest records for testing.

It is especially important for time-series and forecasting problems.
#### Real-World Example
Suppose you have monthly sales data from January 2023 to December 2025.
You divide it into:
- Training: January 2023 – December 2024.
- Validation: January 2025 – June 2025.
- Testing: July 2025 – December 2025.

## Why Preprocessing Must Be Fitted Only on Training Data
Definition
Preprocessing is the process of preparing raw data before giving it to a Machine Learning model.
Some preprocessing techniques need to learn information from the dataset before transforming it.
#### Examples include:
- Missing value imputation
- Standardization
- Normalization
- Categorical encoding
- Feature Selection
- Power Transformation
These techniques must learn their parameters from the training data only.
The learned parameters are then applied to validation and test data.
#### Real-World Example
Suppose a dataset contains employee salaries:
[20000, 30000, 40000, 50000, 60000]
You want to standardize these values.
Standardization calculates the mean and standard deviation.

If you calculate them using the entire dataset, including test records, information from the test data influences preprocessing.
This is a form of data leakage.
Correct Approach
1. Split the dataset.
2. Fit the preprocessing technique on training data.
3. Transform training data.
4. Transform validation data using the same fitted technique.
5. Transform test data using the same fitted technique.
6. Train and evaluate the model.
Understanding fit(), transform(), and fit_transform()
These three methods are extremely important in Machine Learning.
#### --fit()
Learns the required information from the data.
For example, StandardScaler learns:
- Mean
- Standard deviation
#### --transform()
Uses the learned information to transform data.

It does not calculate new mean and standard deviation values.
#### --fit_transform()
Performs both operations:
- Learns the required information.
- Transforms the same data.
