# Feature Selection

## What is Feature Selection?
Definition
Feature Selection is the process of selecting the most useful input features from a dataset for training a Machine Learning model.
It removes unnecessary, irrelevant, or redundant features while retaining features that provide useful information for predicting the target variable.
The main objective is to improve model efficiency, reduce complexity, and sometimes improve prediction performance.
#### Real-World Example
Suppose you want to predict a student's examination marks.

Your dataset contains:
- Study Hours
- Attendance
- Previous Marks
- Sleep Hours
- Student Name
- Phone Number
- Favorite Color
Not every feature is useful for predicting marks.
Study Hours, Attendance, and Previous Marks may be useful, while Student Name and Phone Number generally do not provide meaningful predictive information.
Feature Selection helps identify which features are worth keeping.
#### Business Example
An e-commerce company wants to predict whether a customer will purchase a product.

The dataset contains:
- Customer Age
- Income
- Previous Purchases
- Website Visits
- Product Views
- Customer ID
- Random Tracking Number
Feature Selection helps the company retain useful customer behavior features and remove unnecessary information.
#### AI/ML Use Case
Feature Selection is commonly used in:
- Customer Churn Prediction
- Fraud Detection
- House Price Prediction
- Customer Purchase Prediction
- Medical Diagnosis
- Loan Approval Prediction
- Sales Forecasting

## Why Feature Selection?
Definition
Feature Selection reduces the number of input variables used by a Machine Learning model.
A dataset may contain hundreds or thousands of features, but only some of them may contribute meaningful information to the prediction.
Selecting useful features can make the model simpler, faster, easier to understand, and sometimes more accurate.
Unnecessary features can affect Machine Learning models in several ways.
#### -- Reduced Model Performance
Irrelevant features may introduce noise into the training data.

For example, predicting house prices using:
- House Area
- Location
- Number of Bedrooms
- Wall Color
- Random House ID
Wall Color or Random House ID may not provide meaningful information.
Including such features can sometimes reduce predictive performance.
#### --Overfitting
Overfitting occurs when a model learns patterns specific to the training data rather than general patterns.

When too many irrelevant features are included, certain models may learn accidental relationships between those features and the target.
This can lead to:
- Very good training performance.
- Poor performance on unseen data.
#### -- Increased Training Time
More features mean more computations during training.

For example:
- Dataset A: 10 features
- Dataset B: 1,000 features
Dataset B may require significantly more processing depending on the algorithm and dataset size.
#### --Increased Memory Usage
More features require more memory to store and process the dataset.
This can become important when working with large datasets.
#### --Difficulty Understanding the Model
When a model uses hundreds of features, it becomes difficult to identify which factors actually influence predictions.
Feature Selection can help create a simpler and more interpretable model.
#### --Curse of Dimensionality
As the number of features increases, the feature space becomes larger.
In distance-based algorithms such as KNN, observations can become less distinguishable in high-dimensional spaces, and more data may be required to learn useful patterns.

This problem is known as the Curse of Dimensionality.
Important Note
Feature Selection does not always improve accuracy.
Sometimes a feature that appears weak individually becomes useful when combined with other features. Removing too many features can also reduce model performance

## Irrelevant Features
Definition
Irrelevant Features are input variables that provide little or no useful information about the target variable.
They may contain random information, identifiers, or unrelated attributes.
Removing irrelevant features can reduce noise and simplify model training.
#### Real-World Example
Predicting house prices using:
- House Area
- Location
- Number of Bedrooms
- Random House Number
The Random House Number may not provide meaningful information about the price.
#### Business Example
A bank predicts whether a customer will default on a loan.

Features:
- Income
- Credit Score
- Loan Amount
- Repayment History
- Random Application Reference Number

## Redundant Features
Definition
Redundant Features are features that contain overlapping or nearly identical information.
When multiple features describe essentially the same information, they may unnecessarily increase model complexity.
Removing redundant features can simplify the dataset without losing much useful information.
#### Real-World Example
Suppose a house dataset contains:
- Area in Square Feet
- Area in Square Meters
Both features describe the same house area.
Including both may be redundant.
#### Business Example
A customer dataset contains:
- Monthly Income
- Annual Income
Annual Income is approximately Monthly Income × 12.
Both features provide almost the same information.

```python
import pandas as pd

df = pd.DataFrame({
    "Monthly_Income": [20000, 30000, 40000, 50000],
    "Annual_Income": [240000, 360000, 480000, 600000],
    "Age": [22, 30, 40, 50]
})

print(df.corr())
```

Definition
Correlation-Based Feature Selection identifies relationships between numerical features and the target variable using correlation coefficients.
It can help select features that have a strong linear relationship with the target and identify features that are highly correlated with one another.

Features:
- Website Visits
- Product Views

## Variance Threshold
Definition
Variance Threshold is a feature selection technique that removes features with very low variance.
A feature with almost the same value for every observation generally provides limited information for distinguishing between observations.
Variance Threshold does not require a target variable, making it an unsupervised feature selection technique.
#### Real-World Example
Suppose a dataset contains:
Age = [20, 25, 30, 35, 40]Gender_Code = [1, 1, 1, 1, 1]
Age changes across observations, but Gender_Code is constant.

A constant feature has zero variance and can be removed.
#### Business Example
A company has a customer dataset containing:
- Age
- Income
- Purchase Amount
- Country Code
If every customer in the dataset has the same Country Code, that feature cannot help distinguish customers within that dataset.

```python
import pandas as pd

from sklearn.feature_selection import VarianceThreshold

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Income": [20000, 30000, 40000, 50000, 60000],
    "Constant_Feature": [1, 1, 1, 1, 1]
})

selector = VarianceThreshold(threshold=0)

X_selected = selector.fit_transform(df)

selected_features = df.columns[
    selector.get_support()
]

print("Selected Features:", selected_features)
```

## Univariate Feature Selection
Definition
Univariate Feature Selection evaluates each feature independently against the target variable.
It uses statistical tests to identify features that have a strong relationship with the target.
It is called univariate because each feature is evaluated individually, without considering combinations of features.
#### Real-World Example
Suppose you predict whether a student passes an examination.

Features:
- Study Hours
- Attendance
- Previous Marks
- Student ID
Univariate selection evaluates each feature separately to determine how useful it is for predicting Pass/Fail.
#### Business Example
A bank predicts whether customers will default on loans.
Univariate feature selection can evaluate:
- Income
- Credit Score
- Loan Amount
- Customer Age
It identifies features with strong individual statistical relationships to loan default.
Common Methods
- SelectKBest
- SelectPercentile
- f_classif
- f_regression
- chi2

## Mutual Information
Definition
Mutual Information measures how much information one variable provides about another variable.
It can identify statistical relationships between features and targets, including nonlinear relationships.
Unlike simple correlation, Mutual Information is not limited to linear relationships.
#### Real-World Example
Suppose examination marks depend on study hours in a nonlinear way.
For example, studying from 0 to 4 hours may improve marks significantly, but studying from 8 to 12 hours may produce smaller improvements.
Correlation may not fully capture this relationship.

Mutual Information can detect more general statistical dependence.
#### Business Example
An e-commerce company predicts customer purchases.
A feature such as Product Views may have a nonlinear relationship with Purchase Probability.
Mutual Information can help identify whether Product Views contains useful information about purchases.

## Recursive Feature Elimination (RFE)
Definition
Recursive Feature Elimination is a feature selection technique that repeatedly trains a model and removes the least important features.
It continues eliminating features until the required number of features remains.
RFE uses a Machine Learning estimator to determine which features are important.

#### Real-World Example
Suppose you predict house prices using 20 features.
RFE can repeatedly eliminate less important features until only 5 useful features remain.
#### Business Example
A bank has 100 customer features for predicting loan defaults.

RFE can help identify a smaller set of features that work well with a selected model.

## Feature Importance
Definition
Feature Importance measures how much each feature contributes to a model's predictive performance or decision-making.
It helps identify which input variables are most useful to the trained model.
Feature Importance can be calculated using different techniques depending on the algorithm.
#### Real-World Example
Suppose a house price prediction model uses:
- Area
- Location
- Bedrooms
- Building Age
- Parking Availability
Feature Importance may indicate that Area and Location contribute more strongly to the model's predictions.
#### Business Example
A customer churn model uses:
- Monthly Charges
- Customer Tenure
- Complaints
- Payment History
- Internet Usage
Feature Importance can help the company understand which variables are most influential in predicting churn.
Feature Importance in Random Forest
Random Forest provides a built-in feature importance measure based on the reduction in impurity produced by splits using each feature.

## Permutation Feature Importance
Definition
Permutation Feature Importance measures how much a model's performance decreases when the values of one feature are randomly shuffled.
If shuffling a feature significantly reduces model performance, that feature is likely important for the model.

It can be used with many different trained models.
#### Real-World Example
A house price model uses Area, Location, and Bedrooms.
If randomly shuffling Area significantly reduces prediction accuracy, Area is likely important to the model.
